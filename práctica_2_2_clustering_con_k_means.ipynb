# **K-means**

**K-means (k-medias)** es un algoritmo de aprendizaje automático no supervisado que divide un conjunto de datos en k grupos o clústeres basándose en la similitud de sus características.

Su principal limitación es que el usuario debe definir el valor de k inicialmente.

Este algoritmo se usa cuando tenemos datos sin etiquetas. Por ejemplo, medidas de flores (largo de pétalo, ancho de pétalo) pero no se sabe qué tipos de flores son. Se quiere descubrir cuántos tipos de flores hay.

En realidad, sí se conocen las tres especies, que luego usaremos para comparar el resultado obtenido: Iris **setosa**, Iris **versicolor** e Iris **virginica**.

```python
from sklearn.datasets import load_iris

iris = load_iris()
print(type(iris))
```

```python
datos = iris.data
print(type(datos))
datos
```

```python
# operaciones
primera_flor = datos[0]
primera_flor
un_valor = datos[5, 2]
un_valor
rango10a15 = datos[10:15]
rango10a15
rango0a3 = datos[:3]
rango0a3
```

```python
# se dispone de 150 datos, para cada uno 4 medidas
datos.shape
```

```python
from sklearn.cluster import KMeans

# === este es el número de grupos ===
n_clusters = 3
# =====================================

# se configura el modelo
modelo = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
# se entrena el modelo
modelo.fit(datos)
```

```python
print(modelo.labels_)          # etiqueta de cluster (0, 1, 2...) de cada muestra
print(modelo.cluster_centers_) # coordenadas de los K centroides (shape: K × n_features)
print(modelo.inertia_)         # suma de distancias² (cómo de "compactos" son los clusters)
print(modelo.n_iter_)          # iteraciones que necesitó para converger
```

```python
import matplotlib.pyplot as plt

plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap="viridis")
plt.title(f"Agrupación con {n_clusters} grupos")
plt.show()
```

**K-Means sirve para datos sin etiquetar, ¿qué datos reales llegan sin etiquetar en el mundo profesional?**

Clientes y Comercio (Marketing y E-commerce)

• Historial de compras: Listas de productos, frecuencias de compra y montos gastados por clientes (K-Means los agrupa para descubrir "compradores compulsivos" o "clientes corporativos").
• Navegación web: El rastro de clics (clickstream), tiempo de permanencia en una página y artículos agregados al carrito de un usuario anónimo.

Ciberseguridad y Finanzas

• Tráfico de red: Direcciones IP, puertos utilizados, volumen de bytes transferidos y protocolos. K-Means agrupa el tráfico normal para identificar "grupos" anómalos que representan ciberataques.
• Transacciones bancarias: Logs de transferencias, retiros en cajeros y pagos con tarjeta con variables como hora, monto y país (útil para detectar patrones inusuales de fraude).

Industria, Logística y Operaciones

• Telemetría de maquinaria (IoT): Lecturas de sensores de temperatura, vibración, presión y voltaje en fábricas o turbinas. Los datos llegan en bruto y el algoritmo ayuda a agrupar los estados de funcionamiento "saludables" frente a los "defectuosos".
• Rutas de reparto: Coordenadas GPS, tiempos de entrega y distancias de flotas de transporte para agrupar zonas logísticas eficientes.

Contenido y Redes Sociales

• Documentos de texto: Millones de artículos de noticias, correos de soporte técnico o reseñas de productos sin clasificar. K-Means los agrupa por similitud de palabras para descubrir los temas principales (Topic Modeling).
• Imágenes y audio: Archivos multimedia sin metadatos explicativos (píxeles de fotos para agruparlas por colores dominantes).

Implementa una comparación entre los resultados obtenidos y los resultados reales.

```python
# aquí va la implementación

from sklearn.cluster import KMeans
from sklearn import datasets
import matplotlib.pyplot as plt

# === 1. Introducir / Cargar los datos y las etiquetas reales ===
# (Aquí usamos el dataset Iris como ejemplo clásico, donde 'data' son tus características
# y 'target' son tus especies reales o 'y_real')
iris = datasets.load_iris()
datos = iris.data    # Tus variables de entrada (características)
y_real = iris.target # Tus etiquetas reales (ground truth)

# NOTA: Si tus datos vienen de otro archivo (por ejemplo, un CSV de pandas),
# simplemente reasígnalos así:
# datos = tu_dataframe[['columna1', 'columna2']].values
# y_real = tu_dataframe['columna_etiqueta'].values


# === 2. Bucle para comparar con 2, 4 y 5 grupos ===
lista_clusters = [2, 4, 5]

for n_clusters in lista_clusters:
    # Configurar y entrenar el modelo K-Means
    modelo = KMeans(n_clusters=n_clusters, random_state=0, n_init=10)
    y_pred = modelo.fit_predict(datos)  # Etiquetas predichas por K-Means

    # Crear la figura con dos paneles lado a lado
    fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharex=True, sharey=True)

    # Gráfica izquierda: Especies Reales
    axes[0].scatter(datos[:, 0], datos[:, 1], c=y_real, cmap="viridis", edgecolor='k', s=40)
    axes[0].set_title("Especies Reales", fontsize=12, fontweight='bold')
    axes[0].grid(True, linestyle='--', alpha=0.5)

    # Gráfica derecha: K-Means con n clústeres
    axes[1].scatter(datos[:, 0], datos[:, 1], c=y_pred, cmap="viridis", edgecolor='k', s=40)
    axes[1].set_title(f"K-Means ({n_clusters} clusters)", fontsize=12, fontweight='bold')
    axes[1].grid(True, linestyle='--', alpha=0.5)

    # Título principal de la comparativa
    plt.suptitle(f"Comparativa: Especies reales vs. K-Means ({n_clusters} clústeres)", fontsize=14, fontweight='bold')

    plt.tight_layout()
    plt.show()
```
